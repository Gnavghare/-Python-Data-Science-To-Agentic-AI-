# - 5.Bitwise Operators (Most Imp)


Bitwise Operators are Applicable  on Integerdataonly but not applicable on 
floating point values because they are varying precision.

The purpose of Bitwise Operators is that "Converting Integer data internally into
Binary Format and Performs Binary Operationsin the form Bit by Bit and final result always displayed in the form decimal number system".

Since the operation  are carrying out in the from of bit by bit hence these operator are named as Bitwise Operators.

### In Python programing we have this bitwise operator **

1. Bitwise Left Shift Operator(<<)
2. Bitwise Right Shift Operator(>>)
3. Bitwise 'and' Operator(&)
4. Bitwise 'or' Operator(|)
5. Bitwise complement Operator(~)
6. Bitwise 'xor' Operator(^)


```python
a=10
res=a<<3   # left shift operator will shift the bits of a to the left by 3 positions, effectively multiplying it by 2^3 (or 8). So, res will be 10 * 8 = 80.
print(res)
```

```python
print(12<<3)# left shift operator will shift the bits of 12 to the left by 3 positions, effectively multiplying it by 2^3 (or 8). So, the result will be 12 * 8 = 96.
```

```python
c=a>>3
print(c) # right shift operator will shift the bits of a to the right by 3 positions, effectively dividing it by 2^3 (or 8). So, c will be 10 // 8 = 1.
```

```python
print(12>>3) # right shift operator will shift the bits of 12 to the right by 3 positions, 
# effectively dividing it by 2^3 (or 8). So, the result will be 12 // 8 = 1.
```

```python
print(36>>4) # right shift operator will shift the bits of 36 to the right by 4 positions, effectively dividing it by 2^4 (or 16). So, the result will be 36 // 16 = 2.
```

```python
a=10
b=15
c= a&b
print(c) # bitwise AND operator will perform a bitwise AND operation between the binary representations of a and b. The result will be 10 & 15 = 10 (in binary: 1010 & 1111 = 1010).
```

```python
s1 ={10,20,30}
s2={40,30,60}
s3=s1.intersection(s2)
print(s3)
s4=s1&s2
print(s4) # bitwise AND operator for sets will return the intersection of the two sets. In this case, s1 & s2 will return {30}, which is the common element in both sets.
```

```python
a=10
b=7
c=a|b
print(c) # bitwise OR operator will perform a bitwise OR operation between the binary representations of a and b. The result will be 10 | 7 = 15 (in binary: 1010 | 0111 = 1111).
print(3|4) # bitwise OR operator will perform a bitwise OR operation between the binary representations of 3 and 4. The result will be 3 | 4 = 7 (in binary: 0011 | 0100 = 0111).
```

```python
a=10
b=~a
print(b) # bitwise NOT operator will perform a bitwise NOT operation on the binary representation of a.
# The result will be ~10 = -11 (in binary: ~1010 = 0101).
```

```python
a=4
b=3
c=a^b
print(c) # bitwise XOR operator will perform a bitwise XOR operation between the binary representations
```

# 7. Identity Operator
1. the purpose of the identity Operators is the " to compare the Memory Address of objects"
2. To get Memory Address of an Objects ? variable , we used id().
3. In Python , we have two Identity Operator . They are 

       1. **is**

       2. **is not**

```python
a=None
b=None
print(id(a))
print(id(b))
print(a is b) # it checks whether a and b refer to the same object in memory. Since both a and b are None, they refer to the same object, so the result will be True.
print(a is not b)
```
